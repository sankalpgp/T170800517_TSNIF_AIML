```python
class Dog:
    species = "Canine"  # Class attribute

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def animal_name(self, dog, cat):
        self.dog = dog
        self.cat = cat
        print("Dog name:", self.dog)
        print("Cat name:", self.cat)
        print("Dog name of init fun ", self.name)

    def location(self, state, country):
        self.state = state
        self.country = country
        print("State:", self.state)
        print("Country:", self.country)



# Taking input from user
name = input("Enter dog object name: ")
age = int(input("Enter age: "))

dog_value = input("Enter dog name: ")
cat_value = input("Enter cat name: ")

# Creating object
d = Dog(name, age)
print (d.name)
print (d.age)
print (d.species)
# Calling function with user input
d.animal_name(dog_value, cat_value)

d.location("Maharashtra", "India")

```

```python
class car:
  def __init__(self, brand):
    self.brand = brand

  def show(self, colour):
    self.colour = colour
    print("Car colour : ", self.colour)
    print("Car Brand : ", self.brand)

  def display(self):
     print("Car colour : ", self.colour)
     print("Car Brand : ", self.brand )

c= car("Toyota")
c.show("Red")
c.display()
```

```python
class Details:

  def __init__(self, name, age, height, weight): #called by class name and used default value
    self.name = name
    self.age = age
    self.height = height
    self.weight = weight
    print(f"The name of student is {self.name} and his age is {self.age}")
    print(f"The height of student is {self.height} and weight is {self.weight}" )



  def student(self):
    print(f"The name of student is {self.name} and his age is {self.age}")
    print(f"The height of student is {self.height} and weight is {self.weight}" )

  def person (self, id , roll_no):
    self.id = id
    self.roll_no = roll_no
    print (f"The id of student is {self.id} and roll no is {self.roll_no}and name of that student is {self.name}")




a = Details ("Shubham", 20, 175.67 , 60)
a.student()
a.person(1001, 10)




```
